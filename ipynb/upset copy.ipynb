```r
# ---------------------------------------------------------
# 1. 环境准备与数据加载
# ---------------------------------------------------------
if (!require("pacman")) install.packages("pacman")
pacman::p_load(tidyverse, UpSetR, reticulate)

# 定义莫兰迪色系
morandi_palette <- c("#91A3B0", "#B8AD9E", "#957E6E", "#A5A58D", "#6B705C")

# 12类修饰名称
mod_names <- c('Am', 'Atol', 'Cm', 'Gm', 'Tm', 'Y', 'ac4C', 'm1A', 'm5C', 'm6A', 'm6Am', 'm7G')

# 加载数据
np <- import("numpy")
data_path <- "/home/dc/vscode/vscode20260406/rgcnformer_sum/npy/human3/"
y_12class <- np$load(paste0(data_path, "12loc.npy"))

# 转换为 UpSet 格式的数据框
df_upset <- as.data.frame(y_12class)
colnames(df_upset) <- mod_names

# ---------------------------------------------------------
# 2. 绘图逻辑
# ---------------------------------------------------------
plot_upset_func <- function() {
  # --- 修改2：逻辑实现（过滤小于50的组合） ---
  # 计算所有交集组合的频率
  patterns <- apply(df_upset, 1, paste, collapse = "")
  counts <- table(patterns)
  # 排除全0（不属于任何集合）的行
  none_pattern <- paste(rep("0", ncol(df_upset)), collapse = "")
  counts <- counts[names(counts) != none_pattern]
  # 统计频率 >= 50 的组合个数
  n_valid_intersections <- sum(counts >= 50)

  UpSetR::upset(df_upset, 
        sets = mod_names,
        nintersects = n_valid_intersections, # 配合下方的 order.by="freq" 实现过滤 < 50 的组合
        scale.intersections = "log10",      # 1. 修改：Intersection 柱状图采用对数轴
        set_size.show = TRUE,               # 3. 修改：SetSize 柱状图显示样本数字
        number.angles = 45,                 # 4. 修改：Intersection 柱状图上方数字斜45度
        main.bar.color = morandi_palette[1],
        sets.bar.color = morandi_palette[3],
        matrix.color = morandi_palette[5],
        shade.color = "#E0E0E0",
        order.by = "freq", 
        show.numbers = "yes",               # 确保显示 Intersection 的数字
        text.scale = c(1.5, 1.2, 1.2, 1, 1.5, 1.2),
        point.size = 2.5, 
        line.size = 0.7)
}

# ---------------------------------------------------------
# 3. 保存文件
# ---------------------------------------------------------
# 检查目录是否存在
if(!dir.exists("png4")) dir.create("png4")

pdf("png4/upset_modifications.pdf", width = 12, height = 8)
plot_upset_func()
dev.off()

png("png4/upset_modifications.png", width = 1200, height = 800, res = 120)
plot_upset_func()
dev.off()

cat("UpSet 图绘制完成（已应用对数轴、频率过滤及标签旋转）。\n")
```
