```python
# ---------------------------------------------------------
# 1. 环境准备与数据加载
# ---------------------------------------------------------
if (!require("pacman")) install.packages("pacman")
pacman::p_load(tidyverse, ggplot2, reticulate, showtext)

# 定义莫兰迪色系
morandi_palette <- c("#91A3B0", "#B8AD9E", "#957E6E", "#A5A58D")

# 加载数据
np <- import("numpy")
data_path <- "human3/"
sequences <- np$load(paste0(data_path, "seq.npy"))

# 计算长度 (根据 human.py 逻辑提取每一行的实际列数)
seq_lengths <- data.frame(Length = rep(ncol(sequences), nrow(sequences)))

# ---------------------------------------------------------
# 2. 绘制小提琴图
# ---------------------------------------------------------
p_violin <- ggplot(seq_lengths, aes(x = "Dataset", y = Length)) +
  geom_violin(fill = morandi_palette[2], color = morandi_palette[3], alpha = 0.8) +
  geom_boxplot(width = 0.1, fill = "white", outlier.shape = NA) +
  theme_minimal() +
  labs(title = "RNA Sequence Length Distribution",
       subtitle = "Mer100Dataset Human RNA",
       x = "",
       y = "Nucleotides (bp)") +
  theme(
    plot.title = element_text(size = 16, face = "bold", color = "#4A4A4A"),
    panel.grid.major.x = element_blank(),
    axis.text = element_text(color = "#666666"),
    plot.background = element_rect(fill = "#FDFDFD", color = NA)
  )

# ---------------------------------------------------------
# 3. 保存文件
# ---------------------------------------------------------
ggsave("length_distribution.pdf", p_violin, width = 8, height = 6)
ggsave("length_distribution.png", p_violin, width = 8, height = 6, dpi = 300)

cat("小提琴图绘制完成。\n")
```
