# 分类结果聚类热力图

使用 ComplexHeatmap 绘制带行聚类的性能热力图，展示不同类别之间性能的相似性。

```r
# 加载所需的包
library(ComplexHeatmap)
library(circlize)
library(grid)
library(dplyr)
library(readr)
```

```r
# 读取数据
df <- read.csv("data/res.csv")

# 定义指标顺序
metric_levels <- c("Acc", "Precision", "Recall", "AUC", "F1", "MCC", "AUPRC", "Sn", "Sp")

# 转换为矩阵：行 = Class，列 = 指标
mat <- df %>%
  select(all_of(metric_levels)) %>%
  as.matrix()
rownames(mat) <- df$Class

dim(mat)
print(mat)
```

```r
file.exists("/usr/share/fonts/truetype/msttcorefonts/Times_New_Roman.ttf")
file.exists("/usr/share/fonts/truetype/msttcorefonts/Times_New_Roman_Bold.ttf")
```

```r
library(ComplexHeatmap)
library(circlize)
library(grid)
library(showtext)

# 1. 注册字体：使用简单的别名 "TNR" 避免潜在的空格解析问题
font_add("TNR", 
         regular = "/usr/share/fonts/truetype/msttcorefonts/Times_New_Roman.ttf", 
         bold = "/usr/share/fonts/truetype/msttcorefonts/Times_New_Roman_Bold.ttf")

# 2. 颜色与逻辑设置 (莫兰迪配色)
min_val <- min(mat, na.rm = TRUE)
max_val <- max(mat, na.rm = TRUE)
col_fun = colorRamp2(
  c(min_val, (min_val + max_val) / 2, max_val), 
  c("#B78B80", "#F2E9E4", "#8FA382")
)

get_text_color <- function(value) {
  norm_val <- (value - min_val) / (max_val - min_val)
  ifelse(norm_val > 0.6, "white", "black")
}

# 3. 创建 Heatmap 对象
# 注意：所有的 fontfamily 都指向 "TNR"
ht <- Heatmap(mat, 
        name = "Performance",
        col = col_fun,
        cluster_rows = TRUE,          
        cluster_columns = FALSE,      
        show_row_dend = TRUE,          
        row_dend_gp = gpar(col = "#4A4E69", lwd = 2.0), # 莫兰迪深灰蓝连接线
        row_dend_width = unit(4, "cm"),
        width = unit(3.0, "cm") * ncol(mat),  # 增加宽度适配 18pt
        height = unit(1.5, "cm") * nrow(mat),
        
        column_names_gp = gpar(fontfamily = "TNR", fontface = "bold", fontsize = 18),
        row_names_gp = gpar(fontfamily = "TNR", fontsize = 18),
        
        heatmap_legend_param = list(
          title = "Performance (%)",
          title_gp = gpar(fontfamily = "TNR", fontface = "bold", fontsize = 18),
          labels_gp = gpar(fontfamily = "TNR", fontsize = 18),
          grid_height = unit(1, "cm")
        ),
        
        cell_fun = function(j, i, x, y, width, height, fill) {
          val <- mat[i, j]
          grid.text(sprintf("%.2f", val * 100), x, y, 
                    gp = gpar(fontfamily = "TNR", fontsize = 18, col = get_text_color(val)))
        })

# 4. 保存为 PDF (使用显式渲染块)
pdf("png3/rgcnformer_cls_heatmap_complex2.pdf", width = 18, height = 12)
  showtext_begin() # 关键：强制当前设备使用 showtext
  draw(ht, heatmap_legend_side = "right")
  showtext_end()   # 关键：结束渲染
dev.off()

# 5. 保存为 PNG (高分辨率)
png("png3/rgcnformer_cls_heatmap_complex2.png", width = 18, height = 12, units = "in", res = 300)
  showtext_opts(dpi = 300) # 确保 showtext 知道当前的 DPI
  showtext_begin()
  draw(ht, heatmap_legend_side = "right")
  showtext_end()
dev.off()

# 6. 在 Jupyter/IDE 中显示
showtext_auto()
draw(ht, heatmap_legend_side = "right")
```

## 说明

- **行聚类**：✓ 已开启，性能相似的类别会被聚合在一起，左侧显示树状图
- **列聚类**：✓ 已关闭，保持原始指标顺序
- **颜色映射**：✓ 低值(差性能)红色 #F04B24 → 高值(好性能)绿色 #007A69
- **文字颜色**：✓ 根据背景深浅自动切换，深色背景用白色，浅色背景用黑色
- **数值标注**：✓ 单元格显示 Value × 100，保留两位小数
- **单元格尺寸**：✓ 设置合理宽度和高度
- **树状连接线**：✓ 已添加 `row_dend_gp = gpar(col = "black", lwd = 1)` 可调参数
- **树状图尺寸**：✓ 已添加 `row_dend_width = unit(2, "cm")` 可调树状图宽度


*字体说明：由于系统兼容性问题，这里暂不指定字体，生成后可以在导出的 PDF/PNG 中检查是否为默认字体。如果需要 Times New Roman，可以在导出后用软件修改，或者确认系统字体路径后重新添加字体设置。*

**调整树状连接线**：修改 `row_dend_gp` 参数
- `col = "black"` - 改变连接线颜色，如 `"gray"`, `"#333333"`
- `lwd = 1` - 改变线宽，越大越粗，如 `1.5`, `2`

**调整树状图大小**：修改 `row_dend_width` 参数
- `unit(2, "cm")` - 树状图占用的宽度（垂直方向高度），增大数值让树状图更高
